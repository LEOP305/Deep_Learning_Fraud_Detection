```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

axes[0].bar(["Legitimate", "Fraudulent"], df["Class"].value_counts().values, color=["#2b5c8f", "#d9534f"])
axes[0].set_title("Transaction Class Distribution")
axes[0].set_ylabel("Number of Transactions")

axes[1].hist(df["Amount"], bins=50, color="#2b5c8f", edgecolor="white")
axes[1].set_title("Transaction Amount Distribution")
axes[1].set_xlabel("Amount")
axes[1].set_ylabel("Frequency")

axes[2].boxplot(
    [
        df[df["Class"] == 0]["Amount"],
        df[df["Class"] == 1]["Amount"]
    ],
    labels=["Legitimate", "Fraudulent"],
    patch_artist=True,
    boxprops=dict(facecolor="#e6ecf2")
)
axes[2].set_title("Transaction Amount by Class")
axes[2].set_ylabel("Amount")

plt.tight_layout()
plt.show()
```

```python
duplicates = df[df.duplicated(keep=False)]

print("Total rows involved in duplicates:", len(duplicates))
print("\nClass distribution among duplicated rows:")
print(duplicates["Class"].value_counts())

print("\nClass distribution among duplicated rows (%):")
print(duplicates["Class"].value_counts(normalize=True) * 100)
```

```python
feature_cols = df.columns.drop("Class")

duplicate_groups = (
    df.groupby(list(feature_cols))["Class"]
    .nunique()
)

conflicting_duplicates = duplicate_groups[duplicate_groups > 1]

print("Duplicate groups with conflicting labels:", len(conflicting_duplicates)) # since it's zero, we can safely treat them as redundant rows
```

```python
df = df.drop_duplicates()
```

```python
correlations = df.corr()["Class"].sort_values(ascending=False)

correlations
```

```python
plt.figure(figsize=(10, 8))

correlations.drop("Class").sort_values().plot(kind="barh")

plt.title("Feature Correlation with Fraud")
plt.xlabel("Correlation with Class")
plt.ylabel("Feature")
plt.show()
```

```python
top_features = (
    correlations.drop("Class")
    .abs()
    .sort_values(ascending=False)
    .head(6)
    .index
)

print(top_features)
```

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 8))

for ax, feature in zip(axes.flat, top_features):
    ax.hist(
        df[df["Class"] == 0][feature],
        bins=50,
        alpha=0.5,
        label="Legitimate"
    )
    ax.hist(
        df[df["Class"] == 1][feature],
        bins=50,
        alpha=0.5,
        label="Fraudulent"
    )
    ax.set_title(feature)
    ax.set_xlabel(feature)
    ax.set_ylabel("Frequency")

axes[0, 0].legend()
plt.tight_layout()
plt.show()
```
